# Toy "metagenomics" data set to demo Altair library

```python
import altair as alt
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('data/starwars_species_abundance.txt',sep='\t')
df
```

```python
df[['organism','species_group']].value_counts()
```

```python
df['species_group'].value_counts()
```

```python
# run 1
df1 = df[df['run_id'] == 1]
df1
```

## Stacked bar chart

```python
alt.Chart(df1).mark_bar().encode(
    x = alt.X('timestamp:O'),
    y = 'rel_abund:Q',
    color = 'organism:N',
    #color = 'species_group:N',
    tooltip=['organism:N', 'species_group:N', 'rel_abund:Q', 'timestamp:O', 'run_id:N','site:N','sample_id:N']
).properties(
    title = 'Run 1: Relative Abundance of Species Over Time',
    width = 800,
    height = 400
).interactive()
```

## Area chart

```python
alt.Chart(df1).mark_area().encode(
    x = 'timestamp:T',
    y = 'rel_abund:Q',
    color = 'organism:N',
    #color = 'species_group:N',aaa
    #detail='organism:N',
    tooltip=['organism:N', 'species_group:N', 'rel_abund:Q', 'timestamp:O', 'run_id:N','site:N','sample_id:N']
).properties(
    title = 'Run 1: Relative Abundance of Species Over Time',
    width = 800,
    height = 400
).interactive()
```

## Line chart

```python
alt.Chart(df1).mark_line().encode(
    x = 'timestamp:T',
    y = 'rel_abund:Q',
    color = 'organism:N',
    #color = 'species_group:N',
    detail='organism:N',
    tooltip=['organism:N', 'species_group:N', 'rel_abund:Q', 'timestamp:O', 'run_id:N','site:N','sample_id:N'],
    strokeDash='species_group:N'
).properties(
    title = 'Run 1: Relative Abundance of Species Over Time',
    width = 800,
    height = 400
).interactive()
```

## Multiple runs

```python
alt.Chart(df).mark_line().encode(
    x = 'timestamp:T',
    y = 'rel_abund:Q',
    color = 'organism:N',
    #color = 'species_group:N',aaa
    #detail='run_id:N',
    tooltip=['organism:N', 'species_group:N', 'rel_abund:Q', 'timestamp:O', 'run_id:N','site:N','sample_id:N'],
    column = 'reactor:N',
).properties(
    title = 'Run 1: Relative Abundance of Species Over Time',
    width = 800,
    height = 400
).interactive()
```

## Legend select

```python
selection = alt.selection_point(fields=['organism'], bind='legend')


base = alt.Chart(df).encode(
    x = 'timestamp:T',
    y = 'rel_abund:Q',
    color = 'organism:N',
    #color = 'species_group:N',aaa
    detail='run_id:N',
    tooltip=['organism:N', 'species_group:N', 'rel_abund:Q', 'timestamp:O', 'run_id:N','site:N','sample_id:N'],
).properties(
    title = 'Run 1: Relative Abundance of Species Over Time',
    width = 800,
    height = 200
).interactive()

points = base.mark_point().encode(
    shape='species_group:N',
    size=alt.condition(selection, alt.value(100), alt.value(20))
).add_params(
    selection
)

lines = base.mark_line().encode(
    size=alt.condition(selection, alt.value(2), alt.value(0.5))
).add_params(
    selection
)

chart = points + lines
chart = chart.facet(
    row='reactor:N',
)
chart
```

## mouseover highlight

```python
highlight = alt.selection_point(on='pointerover', fields=['organism'], nearest=True)

base = alt.Chart(df).encode(
    x = 'timestamp:T',
    y = 'rel_abund:Q',
    color = 'organism:N',
    #color = 'species_group:N',aaa
    detail='run_id:N',
    tooltip=['organism:N', 'species_group:N', 'rel_abund:Q', 'timestamp:O', 'run_id:N','site:N','sample_id:N'],
).properties(
    title = 'Run 1: Relative Abundance of Species Over Time',
    width = 800,
    height = 200
).interactive()

points = base.mark_point().encode(
    shape='species_group:N',
    size=alt.condition(highlight, alt.value(100), alt.value(20))
).add_params(
    highlight
)

lines = base.mark_line().encode(
    size=alt.condition(highlight, alt.value(2), alt.value(0.5))
).add_params(
    highlight
)

chart = points + lines
chart = chart.facet(
    row='reactor:N',
)
chart
```
