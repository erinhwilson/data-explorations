# Toy "metagenomics" data set to demo Altair library

```python
import altair as alt
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

import plotly.express as px
import plotly.graph_objects as go
```

```python
df = pd.read_csv('data/starwars_species_abundance.txt',sep='\t')
df
```

## Multiple runs

```python
alt.Chart(df).mark_line().encode(
    x = 'timestamp:T',
    y = 'rel_abund:Q',
    color = 'organism:N',
    #color = 'species_group:N',aaa
    #detail='run_id:N',
    tooltip=['organism:N', 'species_group:N', 'rel_abund:Q', 'timestamp:O', 'run_id:N','site:N','sample_id:N'],
    column = 'reactor:N',
).properties(
    title = 'Run 1: Relative Abundance of Species Over Time',
    width = 400,
    height = 400
).interactive()
```

## Legend select

```python
selection = alt.selection_point(fields=['organism'], bind='legend')


base = alt.Chart(df).encode(
    x = 'timestamp:T',
    y = 'rel_abund:Q',
    color = 'organism:N',
    #color = 'species_group:N',aaa
    detail='run_id:N',
    tooltip=['organism:N', 'species_group:N', 'rel_abund:Q', 'timestamp:O', 'run_id:N','site:N','sample_id:N'],
).properties(
    title = 'Run 1: Relative Abundance of Species Over Time',
    width = 800,
    height = 200
).interactive()

points = base.mark_point().encode(
    shape='species_group:N',
    size=alt.condition(selection, alt.value(100), alt.value(20))
).add_params(
    selection
)

lines = base.mark_line().encode(
    size=alt.condition(selection, alt.value(2), alt.value(0.5))
).add_params(
    selection
)

chart = points + lines
chart = chart.facet(
    row='reactor:N',
)
chart
```

## mouseover highlight

```python
highlight = alt.selection_point(on='pointerover', fields=['organism'], nearest=True)

base = alt.Chart(df).encode(
    x = 'timestamp:T',
    y = 'rel_abund:Q',
    color = 'organism:N',
    #color = 'species_group:N',aaa
    detail='run_id:N',
    tooltip=['organism:N', 'species_group:N', 'rel_abund:Q', 'timestamp:O', 'run_id:N','site:N','sample_id:N'],
).properties(
    title = 'Run 1: Relative Abundance of Species Over Time',
    width = 800,
    height = 200
).interactive()

points = base.mark_point().encode(
    shape='species_group:N',
    size=alt.condition(highlight, alt.value(100), alt.value(20))
).add_params(
    highlight
)

lines = base.mark_line().encode(
    size=alt.condition(highlight, alt.value(2), alt.value(0.5))
).add_params(
    highlight
)

chart = points + lines
chart = chart.facet(
    row='reactor:N',
)
chart
```

## Texture Area chart

```python
site = df1['site'].unique()[0]
run_id = df1['run_id'].unique()[0]

# Tableau 20 color palette for plotly
tableau_20 = [
    "#4E79A7", "#A0CBE8", "#F28E2B", "#FFBE7D","#59A14F", 
    "#8CD17D", "#B6992D", "#F1CE63", "#499894", "#86BCB6",
    "#E15759", "#FF9D9A", "#79706E", "#BAB0AC", "#D37295",
    "#FABFD2", "#B07AA1", "#D4A6C8", "#9D7660", "#D7B5A6",
]

# set column to be used as texture fill
df1['wings'] = df1['species_group'].apply(lambda x: 1 if x == 'flapper' else 0)
    

# plot the stacked area chart
fig = px.area(df1.sort_values('organism'), x="timestamp", y="rel_abund", 
            color="organism",
            color_discrete_sequence=tableau_20,
            hover_data=['organism','rel_abund','sample_id','species_group'],)

# change the y-axis label
fig.update_yaxes(title_text="Relative Abundance (%)")
    
#Loop through the traces and apply pattern fill based on the 'phage' column
for trace in fig.data:
    # Extract the name of the organism for the current trace
    organism_name = trace.name
    
    # Check if 'phage' is True for this organism
    if df1[df1['organism'] == organism_name]['wings'].iloc[0] == True:
        trace.update(fillpattern=dict(shape='x'))

# update the fillcolor of each trace to match the line color (not transparent)
fig.for_each_trace(lambda trace: trace.update(fillcolor = trace.line.color))

# remove background grid
fig.update_layout(
    plot_bgcolor='rgba(0,0,0,0)',
    title=f"{site} - {run_id}",
)

fig.show()
```
